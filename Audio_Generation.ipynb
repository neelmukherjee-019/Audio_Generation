```python
import random
import torch
from IPython.display import Audio, display
from transformers import pipeline, AutoTokenizer, VitsModel

# -----------------------------
# Load models
# -----------------------------
generator = pipeline(
    "text-generation",
    model="distilgpt2",
    device=-1  # Use 0 if you have a GPU
)

tts_model_name = "facebook/mms-tts-eng"

tokenizer = AutoTokenizer.from_pretrained(tts_model_name)
tts_model = VitsModel.from_pretrained(tts_model_name)

# Prompts
prompts = [
    "Artificial intelligence",
    "Once upon a time",
    "The future of science",
    "A beautiful morning",
    "Technology is changing",
]

# Generate and play 5 audio clips
for _ in range(5):

    prompt = random.choice(prompts)

    generated = generator(
        prompt,
        max_new_tokens=25,
        do_sample=True,
        temperature=0.9,
        top_p=0.95,
        pad_token_id=generator.tokenizer.eos_token_id,
    )

    # Extract the generated sentence
    text = generated[0]["generated_text"]

    # Keep only the first sentence
    if "." in text:
        text = text.split(".")[0] + "."

    # Convert to speech
    inputs = tokenizer(text, return_tensors="pt")

    with torch.no_grad():
        waveform = tts_model(**inputs).waveform

    # Play audio
    display(Audio(waveform.squeeze().numpy(),
                  rate=tts_model.config.sampling_rate))
```

```python
import torch
import scipy.io.wavfile as wavfile
from IPython.display import Audio, display
from transformers import AutoTokenizer, VitsModel

# Load model
model_name = "facebook/mms-tts-eng"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = VitsModel.from_pretrained(model_name)

count = 1

while True:

    text = input("\nEnter text (type 'exit' to stop): ")

    if text.lower() == "exit":
        print("Program terminated.")
        break

    inputs = tokenizer(text, return_tensors="pt")

    with torch.no_grad():
        waveform = model(**inputs).waveform

    audio = waveform.squeeze().cpu().numpy()

    filename = f"speech_{count}.wav"

    wavfile.write(
        filename,
        rate=model.config.sampling_rate,
        data=audio
    )

    print(f"Saved: {filename}")

    # Play the generated audio
    display(Audio(audio, rate=model.config.sampling_rate))

    count += 1
```
